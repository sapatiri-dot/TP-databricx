```python
%sql
CREATE DATABASE IF NOT EXISTS compra_venta_autos;
USE compra_venta_autos;
```

```python
%sql
CREATE TABLE MARCAS(
  ID_MARCA BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
  NOMBRE_MARCA VARCHAR(200),
  PAIS_ORIGEN VARCHAR(200),
  ANIO_FUNDACION INT

)
```

```python
%sql
DROP TABLE IF EXISTS MARCAS;
```

```python
%sql
CREATE TABLE VEHICULO(
  ID_VEHICULO INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
  ID_MARCA INT,
  NOMBRE_VEHICULO VARCHAR(200),
  ANIO_FABRICACION INT,
  TIPO_COMBUSTIBLE VARCHAR(200),
  PRECIO DECIMAL(10,2),
  KILOMETRAJE INT,
  FOREIGN KEY (ID_MARCA) REFERENCES MARCAS(ID_MARCA)
);

    
```

```python
%sql
CREATE TABLE CLIENTES(
  ID_CLIENTE INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
  NOMBRE VARCHAR(200),
  APELLIDO VARCHAR(200),
  EMAIL VARCHAR(200),
  TELEFONO DECIMAL (10)
);
```

```python
%sql
CREATE TABLE ORDENES(
  ID_ORDEN BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
  ID_CLIENTE BIGINT,
  ID_VEHICULO BIGINT,
  TIPO_ORDEN VARCHAR(200),
  FECHA DATE,
  FOREIGN KEY (ID_CLIENTE) REFERENCES CLIENTES(ID_CLIENTE),
  FOREIGN KEY (ID_VEHICULO) REFERENCES VEHICULO(ID_VEHICULO)
);
```

```python
%sql
CREATE TABLE PAGO(
  ID_PAGO BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
  ID_ORDEN INT,
  ID_PAGO_DETALLE INT,
  MONTO DECIMAL (10,2),
  FECHA_PAGO DATE,
  METODO_PAGO VARCHAR(200),
  ESTADIO_PAGO VARCHAR(200),
  FOREIGN KEY (ID_ORDEN) REFERENCES ORDENES(ID_ORDEN),
  FOREIGN KEY (ID_PAGO_DETALLE) REFERENCES PAGO_DETALLE(ID_PAGO_DETALLE)
);
```

```python
%sql
CREATE TABLE PAGO_DETALLE(
  ID_PAGO_DETALLE INT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
  ESTADO_PAGO VARCHAR(200) --PENDIENTE/COMPLETADO/CANCELADA
)
  
```

```python
%sql
DROP TABLE IF EXISTS MARCAS;
DROP TABLE IF EXISTS VEHICULO;
DROP TABLE IF EXISTS CLIENTES;
DROP TABLE IF EXISTS ORDENES;
DROP TABLE IF EXISTS PAGO;
DROP TABLE IF EXISTS PAGO_DETALLE;
```
